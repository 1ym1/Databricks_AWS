### Importing Dependancies

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DateType, TimestampType, FloatType
import pyspark.sql.functions as F

catalog_name = 'Ecommerce_BI'
```

##Defining Schemas

##Brands Schema

```python
#Defining the path for raw non delta data
RAW_BRANDS_DATA_PATH = "/Volumes/ecommerce_bi/source_data/raw/brands/*.csv"

#Defining the schema structure for the data file
brand_schema = StructType([
    StructField("brand_code", StringType(), False),
    StructField("brand_name", StringType(), True),
    StructField("category_code", StringType(), True)
])

#Retrieving our data into a Dataframe
df = spark.read.option('header', "true").option("delimiter", ",").schema(brand_schema).csv(RAW_BRANDS_DATA_PATH)
#Adding metadata columns (source file path and ingestion timestamp)
df = df.withColumn("_source_file", F.col("_metadata.file_path")).withColumn("ingested_at", F.current_timestamp())

#Writing the DataFrame as a Delta table to the bronze layer with overwrite mode.
df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(f"{catalog_name}.bronze.brz_brands")

#Display
display(df.limit(5))
```

###Order Items

```python
RAW_ORDERITEM_DATA_PATH = "/Volumes/ecommerce_bi/source_data/raw/order_items/landing/*.csv"

order_items_schema = StructType([
    StructField("dt", StringType(), True),
    StructField("order_ts", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("order_id", StringType(), True),
    StructField("item_seq", StringType(), True),
    StructField("product_id", StringType(), True),
    StructField("quantity", StringType(), True),
    StructField("unit_price_currency", StringType(), True),
    StructField("unit_price", StringType(), True),
    StructField("discount_pct", StringType(), True),
    StructField("tax_amount", StringType(), True),
    StructField("channel", StringType(), True),
    StructField("coupon_code", StringType(), True)
])

df = spark.read.option("header", "true") \
    .option("delimiter",",") \
    .schema(order_items_schema) \
    .csv(RAW_ORDERITEM_DATA_PATH) \
    .withColumn("file_name", F.col("_metadata.file_path")) \
    .withColumn("ingested_at", F.current_timestamp())

df.write.format("delta").mode("overwrite").option("mergeSchema", "True").saveAsTable(f'{catalog_name}.bronze.brz_order_items')

display(df.limit(5))
```
